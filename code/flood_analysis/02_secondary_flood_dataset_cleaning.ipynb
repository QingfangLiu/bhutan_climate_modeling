```python
import pandas as pd
import numpy as np

# Step 1: Load and copy
df2 = pd.read_excel("../../data/Extreme_events_2016_2022.xlsx").copy()

# Step 2: Clean up column names for consistency
df2.rename(columns={
    "Event_Type": "EventType",
    "Location": "Location",
    "Date": "Date",
    "Details": "Description"
}, inplace=True)

# Step 3: Filter to only include flood-related events
df2 = df2[df2["EventType"].str.contains("Flood", case=False)]

# Step 4: Handle date 
df2["Date"] = pd.to_datetime(df2["Date"], errors="coerce")

# Step 5: Add missing columns to match flood_df
df2["District"] = np.nan  # You can fill this manually if you want
df2["Latitude"] = np.nan
df2["Longitude"] = np.nan
df2["Fatalities"] = np.nan  # Optional
df2["Source"] = "Extreme_Events"

# Step 6: Align column order to match flood_df
cols_to_keep = ['Date', 'EventType', 'District', 'Location', 'Latitude', 'Longitude', 'Fatalities', 'Description', 'Source']
df2 = df2[cols_to_keep]
```
