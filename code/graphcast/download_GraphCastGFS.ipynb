## GraphCastGFS Forecast Data Download

The **GraphCast Global Forecast System (GraphCastGFS)** is an experimental forecasting system developed by the National Centers for Environmental Prediction (NCEP) to generate medium-range global forecasts.

- In this notebook, we demonstrate how to directly access and download forecast data from **AWS cloud storage**.

### Key parameters:
- `date`: Forecast initialization date in `YYYYMMDD` format (e.g., `'20250730'`)
- `cycle`: Forecast cycle time (e.g., `'00'` for 00 UTC)
- `forecast_hour`: Forecast lead time (e.g., `'006'` = 6-hour forecast)
- `resolution`: Horizontal resolution string (e.g., `'0p25'` = 0.25°)
- `level_folder`: Directory name on S3 for pressure level datasets (e.g., `'forecasts_13_levels'`)
- `out_dir`: Local directory to save the downloaded file

The filename is constructed based on these parameters and retrieved from the appropriate AWS S3 path. Files are downloaded using the `aws s3 cp` command with `--no-sign-request` for public access.


```python
import os

def download_graphcast_gfs(
    date='20250730',
    cycle='00',
    forecast_hour='006',
    resolution='0p25',
    level_folder='forecasts_13_levels',
    out_dir='../../graphcastgfs_downloads'
):
    # Ensure output directory exists
    os.makedirs(out_dir, exist_ok=True)

    # Construct filename and S3 path
    fname = f"graphcastgfs.t{cycle}z.pgrb2.{resolution}.f{forecast_hour}"
    s3_path = f"s3://noaa-nws-graphcastgfs-pds/graphcastgfs.{date}/{cycle}/{level_folder}/{fname}"

    # Local file path (inside out_dir)
    local_path = os.path.join(out_dir, fname)

    # Download
    cmd = f"aws s3 cp --no-sign-request {s3_path} {local_path}"
    print(f"Downloading: {fname} → {local_path}")
    os.system(cmd)
```

```python
forecast_hours = [f"{h:03d}" for h in range(0, 72+1, 6)]  # e.g., 0 to 72h in 6h steps
    
# 🔁 Example: download multiple forecast hours
for fh in forecast_hours:
    download_graphcast_gfs(date='20250730', cycle='00', forecast_hour=fh, out_dir='../../graphcastgfs_downloads')
```
