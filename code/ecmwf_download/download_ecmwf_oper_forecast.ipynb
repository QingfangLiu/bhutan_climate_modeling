# ECMWF Operational Forecast Downloader

This notebook downloads ECMWF operational forecast data (`.grib2` files)  
from the specified forecast date and run (`00z` or `12z`), using a customizable list of forecast hours.  
It builds file URLs dynamically and saves them to a local folder for further analysis or visualization.

**Details:**
- **Source**: [https://data.ecmwf.int](https://data.ecmwf.int)/forecasts/`{date}`/`{run}z`/ifs/0p25/oper/
- **Format**: GRIB2  
- **Resolution**: 0.25°  
- **Forecast hours**: customizable (e.g., 0, 6, 12, 24...)

**Folder structure reference:**
- `ifs` → *Integrated Forecasting System*, ECMWF’s global NWP model  
- `oper` → *Operational forecast*, ECMWF’s high-resolution official forecast  
- `0p25` → Spatial resolution of 0.25° grid

You can modify the `date`, `run`, and `forecast_hours` in the script to control what gets downloaded.


```python
import os
import requests

# === CONFIGURATION ===
save_dir = "../../ecmwf_forecasts/"
os.makedirs(save_dir, exist_ok=True)

date = "20250727"
run = "00"
base_url = f"https://data.ecmwf.int/forecasts/{date}/{run}z/ifs/0p25/oper"
forecast_hours = [0, 6, 12, 24, 30]  # Add more as needed
```

```python
# === HELPERS ===
def build_filename(date, run, hour):
    return f"{date}{run}0000-{hour}h-oper-fc.grib2"

def download_file(url, output_path):
    print(f"📥 Downloading {url}")
    r = requests.get(url, stream=True)
    if r.status_code == 200:
        with open(output_path, 'wb') as f:
            for chunk in r.iter_content(chunk_size=8192):
                f.write(chunk)
        print(f"✅ Saved to {output_path}")
    else:
        print(f"❌ Failed to download {url} | HTTP {r.status_code}")

# === MAIN LOOP ===
for hour in forecast_hours:
    grib_filename = build_filename(date, run, hour)
    url = f"{base_url}/{grib_filename}"
    output_path = os.path.join(save_dir, grib_filename)

    if not os.path.exists(output_path):
        download_file(url, output_path)
    else:
        print(f"⚠️ Already exists: {grib_filename}")
```
